```python
import sys
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

sys.path.append(r'C:\Users\Acer\Documents\RAISSA\IC')
from Classes.Problem import Problem
from Classes.Solution import Solution
from Classes.Algorithm import VariableNeighborhoodSearch
from Classes.Neighborhood import Swap, Reversion, Insertion, Slide, ETN, RS, SPS, SRPS
```

```python
sets = []
sets.append([Swap(2), Reversion(2), Insertion(2), Slide(2)])
sets.append([ETN(2), RS(2), SPS(2), SRPS(2)])
sets.append([Swap(2), Reversion(2), ETN(2), RS(2)])
```

## Problem size 30

```python
data = Problem()
data.loadFile(r"C:\Users\Acer\Documents\RAISSA\IC\data\data_30.npz")
```

```python
FX_histories = []

for i in range(len(sets)):
    vns = VariableNeighborhoodSearch(sets[i], 100000, 1)
    solution = vns.solve(data)
    FX_history = vns.FX_history
    
    FX_histories.append(FX_history)

FX_histories = [item for sublist in FX_histories for item in sublist]
print(FX_histories)
```

```python
# Calculating the mean and standard deviation
mean_fx = np.mean(FX_histories)
std_fx = np.std(FX_histories)

# Plotting the normal distribution
x = np.linspace(mean_fx - 3*std_fx, mean_fx + 3*std_fx, 1000)
y = stats.norm.pdf(x, mean_fx, std_fx)

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='Normal Distribution')
plt.hist(FX_histories, bins=30, density=True, alpha=0.6, color='g', label='FX_history')

# Adding a specific point to the plot
specific_point = 1.997801471339e+07
percentage = stats.norm.cdf(specific_point, mean_fx, std_fx) * 100

plt.scatter(specific_point, 0, color='red', s=100, zorder=5, label=f"Gurobi's answer: {specific_point}")
plt.axvline(specific_point, color='red', linestyle='--', linewidth=1, zorder=4)
plt.text(specific_point, max(y) / 2, f'{percentage:.2f}%', color='red', ha='center', va='center', fontsize=12, bbox=dict(facecolor='white', alpha=0.6, edgecolor='red'))

plt.title('Normal Distribution of FX_history')
plt.xlabel('FX Value')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```
