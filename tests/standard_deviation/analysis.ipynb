```python
import sys
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
import numpy as np
```

## Problem size 100

```python
filename = '../final_set/solutions_100.npz'
data = np.load(filename)
data = data["solutions"]
a = data[0, :]
b = data[1, :]
c = data[2, :]

solutions = np.concatenate((a, b, c))

mean = np.mean(solutions)
std_dev = np.std(solutions)

gurobi_result = 70598781.54841062

z_score = (gurobi_result - mean) / std_dev

print("Mean:", mean)
print(f"Standard Deviation: {std_dev:.2f}")
print(F"Gurobi Result: {gurobi_result:.2f}")
print(f"Z-Score: {z_score:.2f}")
```

A z-score of approximately *−0.943* means that Gurobi's result is about 0.943 standard deviations below the mean of your experimental results.


Since ∣z∣<1, Gurobi's result is within one standard deviation of the mean of your experimental results.

This suggests that Gurobi's result is well within the range of typical variability observed in your optimization experiments.
