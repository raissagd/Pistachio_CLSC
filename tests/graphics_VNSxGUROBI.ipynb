```python
import numpy as np
import matplotlib.pyplot as plt
```

# VNS x GUROBI

## Instance 100

```python
filename = './tests/run_parallel/results/script_100_fx_values.npz'
data = np.load(filename)
fx_array = data["fx_array"]

# Gurobi's solution
gurobi_sol = 70598781.55

plt.figure(figsize=(8, 6))
plt.boxplot(fx_array)
plt.scatter(1, gurobi_sol, color='red', s=100, label=f"Gurobi's solution: {gurobi_sol}")
plt.ylabel('Solution values')
plt.title('Comparison of 30 runs of the VNS with Gurobi: instance 100')
plt.legend()
plt.show()
```
