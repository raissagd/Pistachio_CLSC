```python
import numpy as np
import pandas as pd
```

```python
def percentage_difference(value1, value2):
    if value1 == value2:
        print("Both values are equal.")
        return
        
    difference = abs(value1 - value2)
    average = (value1 + value2) / 2
    percentage_diff = (difference / average) * 100
        
    if value1 > value2:
        print(f"+{percentage_diff:.2f}%")
    else:
        print(f"-{percentage_diff:.2f}%")
```

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)
```

## Problem size 10

```python
filename = './results/solutions_10_GA.npz'
data = np.load(filename)
data_GA = data["best_solutions"]

avg_GA = calculate_average(data_GA)
print(f"Average GA: {avg_GA:.2f}")
```

```python
filename = './results/solutions_10_VNS.npz'
data = np.load(filename)
data_VNS = data['fx_array']

avg_VNS = calculate_average(data_VNS)
print(f"Average VNS: {avg_VNS:.2f}")
```

```python
percentage_difference(avg_GA, avg_VNS)
```
