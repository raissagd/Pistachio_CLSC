```python
import numpy as np
import pandas as pd
```

```python
def percentage_difference(value1, value2):
    if value1 == value2:
        print("Both values are equal.")
        return
        
    difference = abs(value1 - value2)
    average = (value1 + value2) / 2
    percentage_diff = (difference / average) * 100
        
    if value1 > value2:
        print(f"+{percentage_diff:.2f}%")
    else:
        print(f"-{percentage_diff:.2f}%")
```

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)
```

## Problem size 10

```python
filename = 'solutions_10_GA.npz'
data = np.load(filename)
data_GA = data["best_solutions"]
```

```python
avg_GA = calculate_average(data_GA)
print(f"Average GA: {avg_GA:.2f}")
```

```python
filename = 'solutions_10_VNS.npz'
data = np.load(filename)
data = data["solutions"]
a_VNS = data[0, :]
b_VNS = data[1, :]
c_VNS = data[2, :]
```

```python
avg1_VNS = calculate_average(a_VNS)
avg2_VNS = calculate_average(b_VNS)
avg3_VNS = calculate_average(c_VNS)
    
total_avg_VNS = (avg1_VNS + avg2_VNS + avg3_VNS) / 3
print(f"Average VNS: {total_avg_VNS:.2f}")
```

```python
percentage_difference(avg_GA, total_avg_VNS)
```
