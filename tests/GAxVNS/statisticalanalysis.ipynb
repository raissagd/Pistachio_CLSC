```python
import numpy as np
from scipy import stats
from matplotlib import pyplot as plt
```

```python
ga = np.load('./new_results/GA_100_FX.npz')['fx_array']
vns = np.load('../run_parallel/new_results/script_100_fx_values.npz')['fx_array']
vns = vns[:30]

print(stats.shapiro(ga))
print(stats.shapiro(vns))
print(stats.ttest_ind(ga, vns, equal_var=False, alternative='greater'))
print(stats.mannwhitneyu(ga, vns, alternative='greater'))
```
