```python
def percentage_difference(value1, value2):
    if value1 == value2:
        print("Both values are equal.")
        return
        
    difference = abs(value1 - value2)
    average = (value1 + value2) / 2
    percentage_diff = (difference / average) * 100
        
    if value1 > value2:
        print(f"+{percentage_diff:.2f}%")
    elif value1 < value2:
        print(f"-{percentage_diff:.2f}%")
```

```python
for i in values:
    print(f'Problem {i}:')
    exec(f'percentage_difference(gurobi_{i}, min_a_problem{i})')
    exec(f'percentage_difference(gurobi_{i}, min_b_problem{i})')
```
