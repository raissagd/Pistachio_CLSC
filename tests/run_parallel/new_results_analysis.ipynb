### Results

```python
# Read the CSV file
df = pd.read_csv("./new_results/fx_values.csv")

# Convert columns to numeric, coercing errors (e.g., '-' becomes NaN)
df["Gurobi"] = pd.to_numeric(df["Gurobi"], errors="coerce")
df["VNS"] = pd.to_numeric(df["VNS"], errors="coerce")

# Calculate percentage difference: (VNS - Gurobi) / Gurobi * 100
df["Percentage"] = ((df["VNS"] - df["Gurobi"]) / df["Gurobi"]) * 100

# Format percentage column
df["Percentage"] = df["Percentage"].map(lambda x: f"{x:.2f}%".replace('.', ',') if pd.notna(x) else "-")

# Format numbers using dot as thousands separator and comma as decimal
def format_brazilian(x):
    return f"{x:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")

df["Gurobi"] = df["Gurobi"].map(lambda x: format_brazilian(x) if pd.notna(x) else "-")
df["VNS"] = df["VNS"].map(lambda x: format_brazilian(x) if pd.notna(x) else "-")

# Print as table
print(df.to_string(index=False))
```
