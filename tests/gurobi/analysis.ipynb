```python
filename = 'all_solutions.npz'
data = np.load(filename)
gurobi_values = data["all_solutions"]
print(gurobi_values)

gurobi_10 = gurobi_values[0]
gurobi_30 = gurobi_values[1]
gurobi_100 = gurobi_values[2]
```

```python
def percentage_difference(value1, value2):
    if value1 == value2:
        print("Both values are equal.")
        return
        
    difference = abs(value1 - value2)
    average = (value1 + value2) / 2
    percentage_diff = (difference / average) * 100
        
    if value1 > value2:
        print(f"{value1} is {percentage_diff:.2f}% greater than {value2}.")
    else:
        print(f"{value1} is {percentage_diff:.2f}% less than {value2}.")
```

```python
percentage_difference(gurobi_10, min_a)
percentage_difference(gurobi_10, min_b)
```
