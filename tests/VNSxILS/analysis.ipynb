```python
import sys
sys.path.insert(0, 'Classes/')
import numpy as np
```

```python
def percentage_difference(value1, value2):
    if value1 == value2:
        print("Both values are equal.")
        return
        
    difference = abs(value1 - value2)
    average = (value1 + value2) / 2
    percentage_diff = (difference / average) * 100
        
    if value1 > value2:
        print(f"+{percentage_diff:.2f}%")
    else:
        print(f"-{percentage_diff:.2f}%")
```

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)
```

## Problem size 10

```python
filename = 'solutions_10_ILS.npz'
data = np.load(filename)
data = data["solutions"]
a_ILS = data[0, :]
b_ILS = data[1, :]
c_ILS = data[2, :]
```

```python
avg1_ILS = calculate_average(a_ILS)
avg2_ILS = calculate_average(b_ILS)
avg3_ILS = calculate_average(c_ILS)
    
overall_average_ILS = (avg1_ILS + avg2_ILS + avg3_ILS) / 3
print(overall_average_ILS)
```

```python
filename = 'solutions_10_VNS.npz'
data = np.load(filename)
data = data["solutions"]
a_VNS = data[0, :]
b_VNS = data[1, :]
c_VNS = data[2, :]
```

```python
avg1_VNS = calculate_average(a_VNS)
avg2_VNS = calculate_average(b_VNS)
avg3_VNS = calculate_average(c_VNS)
    
overall_average_VNS = (avg1_VNS + avg2_VNS + avg3_VNS) / 3
print(overall_average_VNS)
```

```python
percentage_difference(overall_average_ILS, overall_average_VNS)
```
