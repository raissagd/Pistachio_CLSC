```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
from scipy.stats import shapiro
plt.rcParams.update({'font.size': 20})
```

# VNS x GA

## Instance 100

```python
filename1 = './GAxVNS/results/solutions_100_VNS.npz'
data1 = np.load(filename1)
fx_array1 = data1["fx_array"]

filename2 = './GAxVNS/results/solutions_100_GA.npz'
data2 = np.load(filename2)
fx_array2 = data2["best_solutions"]

plt.figure(figsize=(10, 6))
plt.boxplot([fx_array1, fx_array2], positions=[1, 2])
plt.ylabel('Solution values')
plt.xticks([1, 2], ['VNS', 'GA'])
plt.title('Comparison between VNS and GA solutions: instance 100')
plt.show()
```

```python
print(shapiro(fx_array1))
print(shapiro(fx_array2))
```

The VNS results are not normally distributed. GA's are.

```python
# Mann-Whitney U test (for non-parametric data)
u_stat, p_value = stats.mannwhitneyu(fx_array1, fx_array2, alternative='less')

print(f'U-statistic: {u_stat}, p-value: {p_value}')
```

The test suggests that the data in fx_array1 tend to be less than the data in fx_array2, and this difference is statistically significant.

## Instance 400

```python
filename1 = './GAxVNS/results/solutions_400_VNS.npz'
data1 = np.load(filename1)
fx_array1 = data1["fx_array"]

filename2 = './GAxVNS/results/solutions_400_GA_FX.npz'
data2 = np.load(filename2)
fx_array2 = data2["fx_array"]

plt.figure(figsize=(10, 6))
plt.boxplot([fx_array1, fx_array2], positions=[1, 2])
plt.ylabel('Solution values')
plt.xticks([1, 2], ['VNS', 'GA'])
plt.title('Comparison between VNS and GA solutions: instance 400')
plt.tight_layout()
plt.show()
```

```python
print(shapiro(fx_array1))
print(shapiro(fx_array2))
```

Normally distributed.

```python
t_stat, p_value = stats.ttest_ind(fx_array1, fx_array2, alternative='less')
print(f't-statistic: {t_stat}, p-value: {p_value}')
```

The mean of fx_array1 is significantly smaller than the mean of fx_array2.
