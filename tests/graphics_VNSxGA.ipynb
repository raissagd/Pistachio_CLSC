```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
from scipy.stats import shapiro
```

# VNS x GA

## Instance 100

```python
filename1 = './tests/GAxVNS/results/solutions_100_VNS.npz'
data1 = np.load(filename1)
fx_array1 = data1["fx_array"]

filename2 = './tests/GAxVNS/results/solutions_100_GA.npz'
data2 = np.load(filename2)
fx_array2 = data2["best_solutions"]

plt.figure(figsize=(10, 6))
plt.boxplot([fx_array1, fx_array2], positions=[1, 2])
plt.ylabel('Solution values')
plt.xticks([1, 2], ['VNS', 'GA'])
plt.title('Comparison between VNS and GA solutions: instance 100')
plt.legend()
plt.show()
```

```python
print(shapiro(fx_array1))
print(shapiro(fx_array2))
```

The VNS results are not normally distributed. GA's are.

```python
# Mann-Whitney U test (for non-parametric data)
u_stat, p_value = stats.mannwhitneyu(fx_array1, fx_array2)

print(f'U-statistic: {u_stat}, p-value: {p_value}')
```

Since the p-value is less than 0.05, there is a significant difference between the two arrays.

```python
if u_stat > 0:
    one_tailed_p_value = p_value / 2  # fx_array_1 > fx_array_2
else:
    one_tailed_p_value = 1 - p_value / 2  # fx_array_2 > fx_array_1

print(f'U-statistic: {u_stat}, one-tailed p-value: {one_tailed_p_value}')
```

fx_array_1 (VNS) tends to have greater values than fx_array_2 (GA).

## Instance 400

```python
filename1 = './tests/GAxVNS/results/solutions_400_VNS.npz'
data1 = np.load(filename1)
fx_array1 = data1["fx_array"]

filename2 = './tests/GAxVNS/results/solutions_400_GA_FX.npz'
data2 = np.load(filename2)
fx_array2 = data2["fx_array"]

plt.figure(figsize=(10, 6))
plt.boxplot([fx_array1, fx_array2], positions=[1, 2])
plt.ylabel('Solution values')
plt.xticks([1, 2], ['VNS', 'GA'])
plt.title('Comparison between VNS and GA solutions: instance 400')
plt.legend()
plt.show()
```

```python
print(shapiro(fx_array1))
print(shapiro(fx_array2))
```

Normally distributed.

```python
t_stat, p_value = stats.ttest_ind(fx_array1, fx_array2)

print(f't-statistic: {t_stat}, p-value: {p_value}')
```

The t-test result indicates a significant difference between the means of the two datasets. 

```python
# Perform the two-tailed t-test
t_stat, p_value = stats.ttest_ind(fx_array1, fx_array2)

# Adjust the p-value for a one-tailed test (divide by 2)
if t_stat > 0:
    one_tailed_p_value = p_value / 2  # fx_array_1 > fx_array_2
else:
    one_tailed_p_value = 1 - p_value / 2  # fx_array_2 > fx_array_1

print(f't-statistic: {t_stat}, one-tailed p-value: {one_tailed_p_value}')
```

fx_array2 has the statistically higher mean.
