```python
def percentage_difference(value1, value2):
    if value1 == value2:
        print("Both values are equal.")
        return
        
    difference = abs(value1 - value2)
    average = (value1 + value2) / 2
    percentage_diff = (difference / average) * 100
        
    if value1 > value2:
        print(f"+{percentage_diff:.2f}%")
    else:
        print(f"-{percentage_diff:.2f}%")
```

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)
```

## Problema de tamanho 100

```python
filename = '../final_set/solutions_100.npz'
data = np.load(filename)
data = data["solutions"]
a = data[0, :]
b = data[1, :]
c = data[2, :]
```

```python
avg1 = calculate_average(a)
avg2 = calculate_average(b)
avg3 = calculate_average(c)
    
overall_average = (avg1 + avg2 + avg3) / 3
print(overall_average)
```

```python
filename = 'solution_100.npz'
data = np.load(filename)
gurobi_val = data["solutions"]

print(gurobi_val)
```

```python
percentage_difference(overall_average, gurobi_val)
```
