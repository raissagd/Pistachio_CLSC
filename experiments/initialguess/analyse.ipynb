```python
import sys
sys.path.append("../../Classes")
from Solution import Solution
from Persistence import PersistMultipleSolutions
from matplotlib import pyplot as plt
import numpy as np

solutions = PersistMultipleSolutions().load(filename="initialguess_400",
                                            filepath="./results/")
```

```python
names = ["Deterministic", "Stochastic"]

num_algorithms = len(solutions)
num_executions = len(solutions[0][0])

fx = np.zeros((num_algorithms, num_executions))
for i in range(num_algorithms):
    for j in range(num_executions):
        fx[i, j] = solutions[i][0][j].FX

plt.boxplot(fx.T, notch=True, patch_artist=True, boxprops=dict(facecolor='lightblue', color='blue'))
plt.xticks(np.arange(1, num_algorithms + 1), names, rotation=45)
plt.ylabel("f(x)")
plt.title("Final function values for different initial guess strategies")
plt.tight_layout()
plt.grid(axis='x')
plt.show()
```
