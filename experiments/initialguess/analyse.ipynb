```python
import sys
sys.path.append("../../Classes")
from Solution import Solution
from Persistence import PersistMultipleSolutions
from utils import boxplot_fx
from matplotlib import pyplot as plt
import numpy as np

solutions = PersistMultipleSolutions().load(filename="initialguess_400",
                                            filepath="./results/")
```

```python
names = ["Deterministic", "Stochastic"]
boxplot_fx(solutions, names)
```

```python
num_executions = len(solutions[0][0])

num_evaluations = [None] * num_executions
fx = [None] * num_executions

for j in range(num_executions):
    convergence = solutions[0][0][j].convergence.get()
    num_evaluations[j] = convergence[0]
    fx[j] = convergence[1]

for i in range(num_executions):
    plt.semilogy(num_evaluations[i], fx[i])
plt.xlabel("Number of function evaluations")
plt.ylabel(r"$f(x)$")
plt.title("Convergence plot for deterministic initial guess")
plt.grid()
plt.show()
```
