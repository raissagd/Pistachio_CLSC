```python
import sys
sys.path.append("../../Classes")
from Solution import Solution
from Persistence import PersistMultipleSolutions, PersistSingleSolution
from utils import boxplot_fx, boxplot_convergence, get_fx_samples
from matplotlib import pyplot as plt
import matplotlib as mpl
import numpy as np

solutions = PersistMultipleSolutions().load(filename="ga_400",
                                            filepath="./results/")
optimum = PersistSingleSolution().load(filename="globopt_400",
                                      filepath="../global_optimum/solutions/")
```

```python
from scipy.stats import kruskal, mannwhitneyu
fx = get_fx_samples(solutions)
print("Kruskal-Wallis Test p-value:")
stat, p = kruskal(*fx)
print(f"{p:.4f}")

print("\nMann-Whitney U Test p-value between Algorithm 0 and 2:")
stat, p = mannwhitneyu(fx[0], fx[2], alternative='greater')
print(f"{p:.4f}")
```
